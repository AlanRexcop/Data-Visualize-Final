```python
# Hàm xử lý chuỗi: xóa khoảng trắng thừa và thay thế theo từ điển
def clean_text(text):
    # Kiểm tra nếu text là NaN hoặc không phải chuỗi thì trả về chính nó
    if not isinstance(text, str):
        return text
    
    text = text.strip()
    for error, correct in corrections.items():
        if error in text:
            return correct
    return text

df['Nhóm hàng'] = df['Nhóm hàng'].apply(clean_text)

# Hiển thị kết quả
print(df[['Năm', 'Nhóm hàng', '1', '2']])

# Lưu file mới với encoding UTF-8 có BOM (để mở bằng Excel không bị lỗi font)
df.to_csv("Long_cleaned"
".csv", encoding="utf-8-sig", index=False)
```
